```python
import config as cfg
import utils.data as data_utils
from utils.model import load_model
from utils.predict import PredictPlants
from utils.validate import ReportGenerator 

MODEL_PATH = cfg.best_model_path 
EXPERIMENT_NAME = "Analysis_of_GatedAttn_v1"

print(f"Configuration Exp Name: {cfg.experiment_name}")

print(f"Loading model from: {MODEL_PATH}")
model = load_model(MODEL_PATH)
predictor = PredictPlants(
    model=model,
    model_name=f"{model.config.get('backbone_name')}_{model.config.get('classifier_head_type')}"
)
print(f"Initialized predictor: {predictor.model_name}")

print("\n--- Loading Datasets ---")
hand_dataloader = data_utils.get_test_dataloader()
_, _, lab_dataloader = data_utils.get_3_dataloaders(batch_size=1)

datasets_to_test = {
    "Hand Test Data": hand_dataloader,
    "Lab Test Data": lab_dataloader
}
print(f"Loaded {len(datasets_to_test)} datasets for comparison.")


# --- 5. Generate the Report ---
report_generator = ReportGenerator(
    predictor=predictor,
    datasets_to_test=datasets_to_test,
    experiment_name=EXPERIMENT_NAME
)

report_generator.generate_pdf_report(
    dataloader_for_per_sample=hand_dataloader
)

# without the per-sample report
# report_generator.generate_pdf_report(
#     dataloader_for_per_sample=None
# )
```
