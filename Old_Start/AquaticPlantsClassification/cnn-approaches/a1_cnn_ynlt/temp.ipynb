```python
import matplotlib as mpl
import shutil

# Find the cache directory
cache_dir = mpl.get_cachedir()

# Delete it
try:
    shutil.rmtree(cache_dir)
    print(f"Successfully deleted Matplotlib cache at: {cache_dir}")
    print("\nIMPORTANT: Please restart your Jupyter kernel now for the changes to take effect.")
    print("In VS Code: Click the 'Restart' button in the toolbar above.")
    print("In Jupyter Lab/Notebook: Go to Kernel > Restart Kernel.")
except OSError as e:
    print(f"Error deleting cache directory {cache_dir}: {e}")
    print("\nPlease try closing all Python processes and running this cell again.")
```
